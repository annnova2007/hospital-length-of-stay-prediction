```python
# ============================================
# SHARED SETUP — same for everyone, do not change
# ============================================
from google.colab import drive
drive.mount('/content/drive')

import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score

df_encoded = pd.read_pickle('/content/drive/MyDrive/ml project/df_encoded.pkl')
df_catboost = pd.read_pickle('/content/drive/MyDrive/ml project/df_catboost.pkl')

RANDOM_STATE = 42
TEST_SIZE = 0.2

# --- For Linear Regression / Random Forest / XGBoost ---
X = df_encoded.drop(columns=['Length of Stay'])
y = df_encoded['Length of Stay']

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=TEST_SIZE, random_state=RANDOM_STATE
)

# --- For CatBoost ---
X_cb = df_catboost.drop(columns=['Length of Stay'])
y_cb = df_catboost['Length of Stay']

X_train_cb, X_test_cb, y_train_cb, y_test_cb = train_test_split(
    X_cb, y_cb, test_size=TEST_SIZE, random_state=RANDOM_STATE
)

print("Standard split:", X_train.shape, X_test.shape)
print("CatBoost split:", X_train_cb.shape, X_test_cb.shape)

def evaluate(y_true, y_pred, model_name):
    rmse = np.sqrt(mean_squared_error(y_true, y_pred))
    mae = mean_absolute_error(y_true, y_pred)
    r2 = r2_score(y_true, y_pred)
    print(f"{model_name} -> RMSE: {rmse:.3f}, MAE: {mae:.3f}, R2: {r2:.3f}")
    return {"model": model_name, "rmse": rmse, "mae": mae, "r2": r2}
```
