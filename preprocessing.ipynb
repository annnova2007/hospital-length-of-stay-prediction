```python
# ============================================
# STEP: Fix Datatypes (corrected)
# ============================================

df = df.copy()

# --- Identifier columns → string ---
id_cols = ['Operating Certificate Number', 'Permanent Facility Id', 'Zip Code']
for col in id_cols:
    df[col] = df[col].astype(str)

# --- Numeric columns ---
df['Length of Stay'] = pd.to_numeric(df['Length of Stay'], errors='coerce')
df['Discharge Year'] = pd.to_numeric(df['Discharge Year'], errors='coerce').astype('int64')
df['Total Charges'] = pd.to_numeric(df['Total Charges'], errors='coerce')
df['Total Costs'] = pd.to_numeric(df['Total Costs'], errors='coerce')

# --- Binary indicator ---
df['Emergency Department Indicator'] = df['Emergency Department Indicator'].map({'Y': 1, 'N': 0}).astype('Int64')

# --- Code columns → category ---
code_cols = ['APR DRG Code', 'APR MDC Code', 'APR Severity of Illness Code']
for col in code_cols:
    df[col] = df[col].astype('category')

# --- Remaining categorical/text columns → category ---
cat_cols = [
    'Health Service Area', 'Hospital County', 'Facility Name', 'Age Group',
    'Gender', 'Race', 'Ethnicity', 'Type of Admission', 'Patient Disposition',
    'CCSR Diagnosis Code', 'CCSR Diagnosis Description',
    'CCSR Procedure Code', 'CCSR Procedure Description',
    'APR DRG Description', 'APR MDC Description',
    'APR Severity of Illness Description', 'APR Risk of Mortality',
    'APR Medical Surgical Description', 'Payment Typology 1'
]
cat_cols = [c for c in cat_cols if c in df.columns]
for col in cat_cols:
    df[col] = df[col].astype('category')

# --- Verify ---
print(df.dtypes)
```

```python
print("Rows before dedup:", len(df))

# Check how many exact duplicate rows exist
dup_count = df.duplicated().sum()
print("Exact duplicate rows found:", dup_count)

# Drop exact duplicates
df = df.drop_duplicates()

print("Rows after dedup:", len(df))
```

```python
# ============================================
# STEP: Encode Categorical Variables (Option A)
# ============================================

# --- Ordinal mappings (apply to both versions, since these are genuinely ordinal) ---
age_map = {
    '0-17': 0,
    '18-29': 1,
    '30-49': 2,
    '50-69': 3,
    '70 or Older': 4
}

severity_map = {
    'Minor': 0,
    'Moderate': 1,
    'Major': 2,
    'Extreme': 3
}

df['Age Group'] = df['Age Group'].map(age_map)
df['APR Severity of Illness Description'] = df['APR Severity of Illness Description'].map(severity_map)
df['APR Risk of Mortality'] = df['APR Risk of Mortality'].map(severity_map)

# ============================================
# BRANCH 1: df_encoded  →  Linear Regression, Random Forest, XGBoost
# ============================================

df_encoded = df.copy()

# Low-cardinality nominal → one-hot
low_card_cols = [
    'Gender', 'Race', 'Ethnicity', 'Type of Admission',
    'Patient Disposition', 'APR Medical Surgical Description',
    'Payment Typology 1'
]
low_card_cols = [c for c in low_card_cols if c in df_encoded.columns]
df_encoded = pd.get_dummies(df_encoded, columns=low_card_cols, drop_first=True)

# High-cardinality nominal → frequency encoding
high_card_cols = [
    'Health Service Area', 'Hospital County', 'Facility Name',
    'CCSR Diagnosis Code', 'CCSR Diagnosis Description',
    'CCSR Procedure Code', 'CCSR Procedure Description',
    'APR DRG Description', 'APR MDC Description',
    'Zip Code'
]
high_card_cols = [c for c in high_card_cols if c in df_encoded.columns]

for col in high_card_cols:
    freq = df_encoded[col].value_counts(normalize=True)
    df_encoded[col + '_freq'] = df_encoded[col].map(freq)

df_encoded = df_encoded.drop(columns=high_card_cols)

# APR code columns → keep as category, but Linear Reg/RF/XGB(sklearn) need numeric
# so cast to int for these three models
code_cols = ['APR DRG Code', 'APR MDC Code', 'APR Severity of Illness Code']
for col in code_cols:
    df_encoded[col] = df_encoded[col].astype(int)

print("df_encoded shape:", df_encoded.shape)
print(df_encoded.dtypes)

# ============================================
# BRANCH 2: df_catboost  →  CatBoost (native categorical handling)
# ============================================

df_catboost = df.copy()

# CatBoost needs categorical columns as string or category dtype (not float/object mix)
cat_features = [
    'Health Service Area', 'Hospital County', 'Facility Name',
    'Gender', 'Race', 'Ethnicity', 'Type of Admission', 'Patient Disposition',
    'CCSR Diagnosis Code', 'CCSR Diagnosis Description',
    'CCSR Procedure Code', 'CCSR Procedure Description',
    'APR DRG Description', 'APR MDC Description',
    'APR Medical Surgical Description', 'Payment Typology 1',
    'Zip Code', 'APR DRG Code', 'APR MDC Code', 'APR Severity of Illness Code'
]
cat_features = [c for c in cat_features if c in df_catboost.columns]

for col in cat_features:
    df_catboost[col] = df_catboost[col].astype(str)  # CatBoost wants string, not pandas category

print("df_catboost shape:", df_catboost.shape)
print("Categorical features for CatBoost:", cat_features)
```

```python
print(len(df_encoded), len(df_catboost))
```

```python
# ============================================
# STEP: Redo — Drop leakage columns, then handle LOS outliers properly
# ============================================

# --- 1. Drop leakage columns (known only after/during stay) ---
leak_cols = ['Total Charges', 'Total Costs']
df_encoded = df_encoded.drop(columns=[c for c in leak_cols if c in df_encoded.columns])
df_catboost = df_catboost.drop(columns=[c for c in leak_cols if c in df_catboost.columns])

# --- 2. Cap (winsorize) Length of Stay at 1st/99th percentile instead of dropping rows ---
lower_cap = df_encoded['Length of Stay'].quantile(0.01)
upper_cap = df_encoded['Length of Stay'].quantile(0.99)
print(f"Length of Stay caps: lower={lower_cap}, upper={upper_cap}")

df_encoded['Length of Stay'] = df_encoded['Length of Stay'].clip(lower_cap, upper_cap)
df_catboost['Length of Stay'] = df_catboost['Length of Stay'].clip(lower_cap, upper_cap)

print("Rows retained:", len(df_encoded), len(df_catboost))
print(df_encoded['Length of Stay'].describe())
```

```python
# Drop rows where target is missing
df_encoded = df_encoded.dropna(subset=['Length of Stay']).reset_index(drop=True)
df_catboost = df_catboost.dropna(subset=['Length of Stay']).reset_index(drop=True)

print(len(df_encoded), len(df_catboost))
```
